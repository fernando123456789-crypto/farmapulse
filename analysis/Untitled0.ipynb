```python
import pandas as pd

df = pd.read_csv('scrapes_log_rows.csv')

# 1. Tasa de éxito general del scraping (%)
total_ejecuciones = len(df)
exitosos = df['encontrado'].sum()
tasa_exito = (exitosos / total_ejecuciones) * 100
print(f"Tasa de éxito global: {tasa_exito:.2f}%")

# 2. ¿Cuáles farmacias tienen más fallos (donde 'encontrado' es False)?
fallos = df[df['encontrado'] == False]
print("\n--- Farmacias con más registros sin resultados (Fallos) ---")
print(fallos['farmacia_id'].value_counts())

# 3. Promedio de cantidad de resultados cuando SÍ se encuentra el medicamento
promedio_resultados = df[df['encontrado'] == True]['cantidad_resultados'].mean()
print(f"\nPromedio de resultados encontrados por búsqueda exitosa: {promedio_resultados:.1f}")
```
